### Core Theoretical Principles:
**LightGBM (Ke et al., 2017)** achieves order-of-magnitude training speedups with zero accuracy degradation via two algorithmic breakthroughs:
1. **GOSS (Gradient-based One-Side Sampling)**: Retains instances with large gradients $|g_i| > a$ and randomly samples small gradient instances with probability $b$, reweighting by $(1-a)/b$:
$$\tilde{g}_i = \begin{cases} g_i & \text{if } i \in A \\ \frac{1-a}{b} g_i & \text{if } i \in B \end{cases}$$
2. **EFB (Exclusive Feature Bundling)**: Bundles mutually exclusive sparse features into dense feature histograms.
3. **Leaf-Wise (Best-First) Tree Growth**: Splits the leaf with maximum delta loss $\Delta \mathcal{L}$, achieving significantly deeper asymmetry and lower loss than depth-wise schemes.


```python
# STEP 1: ENTERPRISE ENVIRONMENT & REPRODUCIBILITY SEED
import os
import time
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report, confusion_matrix
from lightgbm import LGBMClassifier

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print("✅ STEP 1: Titanic LightGBM Classifier Environment initialized.")
```

```python
# STEP 2: DATA INGESTION & VALIDATION SPLIT
data_path = 'data/titanic/Titanic-Dataset.csv' if os.path.exists('data/titanic/Titanic-Dataset.csv') else 'data/titanic/titanic.csv'
df = pd.read_csv(data_path)
df.columns = [c.strip().lower() for c in df.columns]

target = 'survived'
num_cols = ['pclass', 'age', 'sibsp', 'parch', 'fare']
cat_cols = ['sex', 'embarked']

X = df[num_cols + cat_cols]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print(f"Training shape: {X_train.shape} | Testing shape: {X_test.shape}")
df.head()
```

```python
# STEP 3: PREPROCESSING ARCHITECTURE
num_pipe = Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())])
cat_pipe = Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('ohe', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))])

preprocessor = ColumnTransformer([
    ('num', num_pipe, num_cols),
    ('cat', cat_pipe, cat_cols)
])
```

```python
# STEP 4: LIGHTGBM PIPELINE ARCHITECTURE
lgb_pipe = Pipeline([
    ('prep', preprocessor),
    ('lgb', LGBMClassifier(
        n_estimators=100,
        learning_rate=0.05,
        num_leaves=31,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=42,
        verbose=-1,
        n_jobs=4
    ))
])
```

```python
# STEP 5 & 6: TRAINING PIPELINE
lgb_pipe.fit(X_train, y_train)
lgb_model = lgb_pipe.named_steps['lgb']
print(f"LightGBM Classifier fitted successfully.")
```

```python
# STEP 7: EVALUATION ON TEST SET
y_pred = lgb_pipe.predict(X_test)
y_probs = lgb_pipe.predict_proba(X_test)[:, 1]

acc = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_probs)

print(f"Test Accuracy: {acc:.4f} | F1: {f1:.4f} | ROC AUC: {auc:.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
# STEP 8: FEATURE IMPORTANCE
ohe = preprocessor.named_transformers_['cat'].named_steps['ohe']
cat_names = list(ohe.get_feature_names_out(cat_cols))
feature_names = num_cols + cat_names

importances = lgb_model.feature_importances_
fi_df = pd.DataFrame({'feature': feature_names, 'importance': importances}).sort_values('importance', ascending=False)

plt.figure(figsize=(8, 4))
sns.barplot(data=fi_df, x='importance', y='feature', palette='Blues_r')
plt.title('LightGBM Titanic Feature Importances (Split Counts)')
plt.show()
```

```python
# STEP 9: 10-FOLD CV STABILITY
scores = cross_val_score(lgb_pipe, X, y, cv=10, scoring='roc_auc')
print(f"10-Fold CV ROC AUC: {scores.mean():.4f} +/- {scores.std():.4f}")
```

```python
# STEP 10: CONFUSION MATRIX
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Perished', 'Survived'], yticklabels=['Perished', 'Survived'])
plt.title('LightGBM Titanic Confusion Matrix')
plt.show()
```

```python
# STEP 11: PRODUCTION SERIALIZATION
os.makedirs('production_models', exist_ok=True)
model_path = 'production_models/titanic_lgbm_pipeline.joblib'
joblib.dump(lgb_pipe, model_path, compress=3)
print(f"Model saved to: {model_path}")
```

```python
# STEP 12: REAL-TIME LATENCY SMOKE TEST
loaded = joblib.load(model_path)
sample = X_test.iloc[[0]]
for _ in range(10): _ = loaded.predict(sample)

lats = []
for _ in range(100):
    t0 = time.perf_counter()
    loaded.predict(sample)
    lats.append((time.perf_counter() - t0) * 1000)
avg_lat = np.mean(lats)
print(f"Mean Latency: {avg_lat:.4f} ms | P99: {np.percentile(lats, 99):.4f} ms")
assert avg_lat < 50.0, "Latency SLA violated"
print("✅ Production SLA Passed (< 50.0 ms)")
```
