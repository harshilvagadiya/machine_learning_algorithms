### Core Theoretical Principles:
Demonstrates **validation-based early stopping** (`validation_fraction=0.15`, `n_iter_no_change=10`) and high-cardinality multi-contract preprocessor design with `ColumnTransformer`.


```python
# STEP 1: ENTERPRISE ENVIRONMENT SETUP
import os
import time
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, f1_score

np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print("✅ STEP 1: Retention Analytics Environment ready.")
```

## 📥 STEP 2 & 3: INGESTION & DATASET HARMONIZATION
Loading Kaggle Telco Customer Churn dataset (7,043 subscriber accounts).


```python
# STEP 2 & 3: INGESTION & AUDIT
data_path = 'data/customer_churn/WA_Fn-UseC_-Telco-Customer-Churn.csv'
df = pd.read_csv(data_path)

# Handle whitespace in TotalCharges
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].astype(str).str.strip(), errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

print(f"📊 Dataset Shape: {df.shape[0]} subscribers, {df.shape[1]} attributes")
print(f"Churn Distribution:\n{df['Churn'].value_counts(normalize=True).round(3)}")
df.head(3)
```

## 🧹 STEP 4 & 5: SEGREGATION & HYGIENE
Separating features and target `Churn`.


```python
# STEP 4 & 5: SEGREGATION
X = df.drop(columns=['customerID', 'Churn']).copy()
y = (df['Churn'] == 'Yes').astype(int)

print(f"Subscriber Features: {X.shape[1]}")
print(f"Missing attributes: {X.isnull().sum().sum()}")
```

## 🔒 STEP 6: STRATIFIED SPLIT
80/20 stratified split.


```python
# STEP 6: SPLIT
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)
print(f"Train Subscribers: {X_train.shape[0]}")
print(f"Test Subscribers : {X_test.shape[0]}")
```

## ⚙️ STEP 7: ROBUST COLUMNTRANSFORMER
Encoding categorical plan attributes and standardizing continuous tenures.


```python
# STEP 7: PREPROCESSOR
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
cat_cols = [c for c in X.columns if c not in num_cols]

preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), num_cols),
    ('cat', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('ohe', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False))
    ]), cat_cols)
])
print("✅ STEP 7: Preprocessor ready.")
```

## ⚡ STEP 8: EARLY STOPPING DEMONSTRATION
Training GradientBoostingClassifier with internal validation-based early stopping.


```python
# STEP 8: EARLY STOPPING FIT
gbc_early = Pipeline([
    ('prep', preprocessor),
    ('gbc', GradientBoostingClassifier(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=3,
        subsample=0.8,
        validation_fraction=0.15,
        n_iter_no_change=10,
        tol=1e-4,
        random_state=42
    ))
])
gbc_early.fit(X_train, y_train)

actual_trees = gbc_early.named_steps['gbc'].n_estimators_
print("="*65)
print(f"Configured Max Trees   : 500")
print(f"Early Stopped At Tree  : {actual_trees} (Saved {500 - actual_trees} redundant tree iterations!)")
print(f"Test ROC-AUC Score     : {roc_auc_score(y_test, gbc_early.predict_proba(X_test)[:, 1]):.4f}")
print("="*65)
```

## 🎯 STEP 9: TUNING HYPERPARAMETERS VIA 3-FOLD CV
Optimizing `learning_rate` and `max_depth`.


```python
# STEP 9: GRID SEARCH
param_grid = {
    'gbc__n_estimators': [100, 150],
    'gbc__learning_rate': [0.03, 0.08],
    'gbc__max_depth': [3, 4]
}

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
pipe_tune = Pipeline([
    ('prep', preprocessor),
    ('gbc', GradientBoostingClassifier(subsample=0.8, random_state=42))
])

grid = GridSearchCV(pipe_tune, param_grid=param_grid, cv=cv, scoring='roc_auc', n_jobs=-1)
grid.fit(X_train, y_train)

best_model = grid.best_estimator_
print(f"🏆 Best Retention Hyperparameters: {grid.best_params_}")
print(f"🎯 Best 3-Fold CV ROC-AUC: {grid.best_score_*100:.2f}%")
```

## 📊 STEP 10 & 11: RETENTION REPORT & RESIDUALS
Evaluating retention intervention thresholds.


```python
# STEP 10 & 11: EVALUATION REPORT
y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

print("📋 RETENTION CAMPAIGN EVALUATION REPORT:")
print(classification_report(y_test, y_pred, target_names=['Retained', 'Churn Risk']))
print(f"Test ROC-AUC Score : {roc_auc_score(y_test, y_prob):.4f}")
print(f"Test F1-Score      : {f1_score(y_test, y_pred):.4f}")
```

## 💾 STEP 12 & 13: PRODUCTION SERIALIZATION & RETENTION SMOKE TEST
Deploying pipeline and validating real-time customer account churn assessment latency.


```python
# STEP 12 & 13: SERIALIZATION & SMOKE TEST
os.makedirs('production_models', exist_ok=True)
model_path = 'production_models/telco_churn_gbc_pipeline.joblib'
joblib.dump(best_model, model_path, compress=3)
print(f"💾 Production pipeline saved to: {model_path} ({os.path.getsize(model_path):,} bytes)")

loaded_pipe = joblib.load(model_path)
sample_account = X_test.iloc[[0]]

t0 = time.perf_counter()
pred_churn = loaded_pipe.predict(sample_account)[0]
pred_prob = loaded_pipe.predict_proba(sample_account)[0, 1]
latency_ms = (time.perf_counter() - t0) * 1000

print(f"\n📱 LIVE SUBSCRIBER RISK EVALUATION:")
print(f"   Retention Warning : {'🚨 HIGH CHURN RISK' if pred_churn == 1 else '🛡️ SECURE SUBSCRIBER'}")
print(f"   Churn Probability : {pred_prob * 100:.2f}%")
print(f"   Evaluation Latency: {latency_ms:.3f} ms (SLA < 2.0ms: {'PASS' if latency_ms < 2.0 else 'CHECK'})")
```
